## Data Understanding – Key Observations

1. FX prices are non-stationary while returns are stationary.
2. Volatility clustering is observed.
3. News volume spikes correlate with increased volatility.
4. Macro indicators show strong interdependencies.
5. Economic events are sparse but high-impact.

## Architectural Implication

This justifies a multi-agent architecture:

- Technical Agent → InfluxDB high-frequency data
- Macro Agent → PostgreSQL economic indicators
- Sentiment Agent → PostgreSQL news data
- Event Agent → PostgreSQL economic calendar

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
from statsmodels.tsa.stattools import adfuller
from pandas.plotting import autocorrelation_plot
from influxdb_client import InfluxDBClient
import os
from dotenv import load_dotenv

%matplotlib inline
load_dotenv()
```

=====================================================
🔵 SECTION 1 – MARKET DATA (InfluxDB)
=====================================================

Connexion InfluxDB

```python
client = InfluxDBClient(
    url=os.getenv("INFLUXDB_URL"),
    token=os.getenv("INFLUXDB_TOKEN"),
    org=os.getenv("INFLUXDB_ORG")
)

query_api = client.query_api()
```

Récupération des données: 
On récupère 180 jours pour analyse

```python
bucket = os.getenv("INFLUXDB_BUCKET")

query = f'''
from(bucket: "{bucket}")
  |> range(start: -180d)
  |> filter(fn: (r) => r["_measurement"] == "forex_prices")
  |> filter(fn: (r) => r["_field"] == "open" or r["_field"] == "high" or r["_field"] == "low" or r["_field"] == "close" or r["_field"] == "volume")
  |> pivot(rowKey:["_time"], columnKey: ["_field"], valueColumn: "_value")
'''

df_px = query_api.query_data_frame(query)
```

```python
if isinstance(df_px, list):
    df_px = pd.concat(df_px, ignore_index=True)

df_px.head()
```

```python
df_px["_time"] = pd.to_datetime(df_px["_time"])
df_px = df_px.sort_values("_time").set_index("_time")

print(df_px.columns)
df_px.head()
```

Restructuration des données
InfluxDB retourne format long → pivot nécessaire

```python
df_eur = df_px[(df_px["symbol"]=="EURUSD") & (df_px["timeframe"]=="1H")].copy()

df_eur["close"] = df_eur["close"].astype(float)
df_eur["log_ret"] = (pd.Series(df_eur["close"]).apply(lambda x: float(x))).pipe(np.log).diff()

df_eur.head()
```

Price Chart

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure()
df_eur["close"].plot()
plt.title("EURUSD Close (1H)")
plt.show()
```

Log Returns

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure()
df_eur["close"].plot()
plt.title("EURUSD Close (1H)")
plt.show()
```

Distribution des Returns

```python
plt.figure()
df_eur["log_ret"].dropna().plot(kind="hist", bins=100)
plt.title("Distribution of Log Returns")
plt.show()
```

Rolling volatility

```python
df_eur["volatility"] = df_eur["log_ret"].rolling(50).std()

plt.figure()
df_eur["volatility"].dropna().plot()
plt.title("Rolling Volatility (50)")
plt.show()
```

Stationarity Test

```python
from statsmodels.tsa.stattools import adfuller

p_price = adfuller(df_eur["close"].dropna())[1]
p_ret = adfuller(df_eur["log_ret"].dropna())[1]

print("Price p-value:", p_price)
print("Returns p-value:", p_ret)
```

=====================================================
🟣 SECTION 2 – MACRO DATA (PostgreSQL)
=====================================================

```python
engine = create_engine(
    "postgresql+psycopg2://forex_user:forex_pass_2026@localhost:5432/forex_metadata"
)

df_macro = pd.read_sql("SELECT * FROM economic_indicators", engine)
```

Parsing & Pivot

```python
df_macro["date"] = pd.to_datetime(df_macro["date"])
df_macro = df_macro.sort_values("date")

macro_pivot = df_macro.pivot_table(
    values="value",
    index="date",
    columns="series_id"
)
```

Correlation Heatmap

```python
plt.figure()
plt.imshow(macro_pivot.corr(), aspect='auto')
plt.colorbar()
plt.title("Macro Correlation Matrix")
plt.show()
```

=====================================================
🟡 SECTION 3 – NEWS DATA
=====================================================

```python
df_news = pd.read_sql("SELECT * FROM news_articles", engine)
df_news["published_at"] = pd.to_datetime(df_news["published_at"])
```

Volume Journalier

```python
news_daily = df_news.set_index("published_at").resample("D").size()

plt.figure()
news_daily.plot()
plt.title("Daily News Volume")
plt.show()
```

Distribution Longueur Articles

```python
df_news["text_length"] = df_news["content"].fillna("").astype(str).str.len()

plt.figure()
df_news["text_length"].plot(kind="hist", bins=50)
plt.title("Article Length Distribution")
plt.show()
```

=====================================================
🟠 SECTION 4 – ECONOMIC EVENTS
=====================================================

```python
df_events = pd.read_sql("SELECT * FROM economic_events", engine)
df_events["event_date"] = pd.to_datetime(df_events["event_date"])
```

Events par Devise

```python
if not df_events.empty:
    plt.figure()
    df_events["currency"].value_counts().plot(kind="bar")
    plt.title("Events per Currency")
    plt.show()
else:
    print("⚠️ No economic events data available.")
```

=====================================================
🔥 SECTION 5 – MULTI-MODAL ALIGNMENT
=====================================================

Align Market Volatility & News

```python
vol_hourly = df_eur["volatility"].resample("D").mean()

combined = pd.DataFrame({
    "volatility": vol_hourly,
    "news_count": news_daily
}).fillna(0)

plt.figure()
combined.plot(subplots=True)
plt.show()
```
